4. busca de cliente 

```python
def buscar_cliente(nome, lista_clientes):
    for i in range(len(lista_clientes)):
        if lista_clientes[i] == nome:
            return i
    return -1

# Teste
clientes = ["Ana", "Carlos", "Beatriz", "Mariana", "João"]
print("Mariana está no índice:", buscar_cliente("Mariana", clientes))
print("Lucas está no índice:", buscar_cliente("Lucas", clientes))
```

5. Contar Ocorrências

```python
def conta_ocorrencias(valor, lista):
    contador = 0
    for elemento in lista:
        if elemento == valor:
            contador += 1
    return contador

# Teste
print("O valor 3 aparece:", conta_ocorrencias(3, [1, 3, 5, 3, 7]), "vezes.")
```

6. Remover Elemento (Deslizamento)

```python
def remove_elemento(lista, valor):
    posicao = -1
    for i in range(len(lista)):
        if lista[i] == valor:
            posicao = i
            break
            
    if posicao == -1:
        return None
    
    removido = lista[posicao]
    
    # Desliza os elementos para a esquerda
    for i in range(posicao, len(lista) - 1):
        lista[i] = lista[i + 1]
        
    lista.pop() # Remove a sobra no final
    return removido

# Teste
numeros = [10, 25, 30, 45, 50]
removido = remove_elemento(numeros, 30)
print("Removido:", removido, "| Nova lista:", numeros)
```

Projeto 1: Encontrando seu Filme

```python
def buscar_filme(titulo_procurado, catalogo):
    for i in range(len(catalogo)):
        if catalogo[i].lower() == titulo_procurado.lower():
            return i
    return -1

catalogo_filmes = ["Inception", "Interstellar", "Matrix"]
filme = input("Qual filme deseja buscar? ")
pos = buscar_filme(filme, catalogo_filmes)

if pos != -1:
    print(f"Filme encontrado! Posição: {pos}")
else:
    print("Filme não está no catálogo.")
```
