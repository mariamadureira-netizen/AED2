```python
ef busca_binaria(lista, alvo):
 esquerda = 0
 direita = len(lista) - 1
 while esquerda <= direita:
 meio = (esquerda + direita) // 2
 if lista[meio] == alvo:
 return meio
 elif lista[meio] < alvo:
 esquerda = meio + 1
 else:
 direita = meio - 1
 return -1
```

Exercício 6 

```python
def busca_linear(lista, numero):
    for elemento in lista:
        if elemento == numero:
            return True
    return False


lista = [5, 2, 8, 1, 9, 3]

print(busca_linear(lista, 8))  # True
print(busca_linear(lista, 7))  # False
```

a busca linear tem complexidade O(n), enquanto a busca utilizando dicionário tem O(1) por consulta, após o pré-processamento O(n).

Projeto 1 

```python
def buscar_aluno_linear(alunos, cpf):
    for aluno in alunos:
        if aluno["cpf"] == cpf:
            return aluno
    return None


def buscar_aluno_eficiente(alunos, cpf):
    # Pré-processamento: cria um índice usando o CPF
    indice = {}

    for aluno in alunos:
        indice[aluno["cpf"]] = aluno

    # Busca pelo CPF
    return indice.get(cpf)


# Lista de ex-alunos
alunos = [
    {"nome": "João Silva", "cpf": "111.111.111-11", "curso": "ADS"},
    {"nome": "Maria Souza", "cpf": "222.222.222-22", "curso": "Gestão Empresarial"},
    {"nome": "Carlos Oliveira", "cpf": "333.333.333-33", "curso": "ADS"},
    {"nome": "Ana Santos", "cpf": "444.444.444-44", "curso": "Gestão Empresarial"}
]

# Testando busca linear
resultado = buscar_aluno_linear(alunos, "333.333.333-33")
print("Busca linear:", resultado)

# Testando busca eficiente
resultado = buscar_aluno_eficiente(alunos, "333.333.333-33")
print("Busca eficiente:", resultado)
```

```python
def recomendar_disciplinas(historicos, aluno_original):
    recomendacoes = []

    for historico in historicos:
        # Verifica se existe pelo menos uma disciplina em comum
        if any(disciplina in aluno_original for disciplina in historico):
            
            # Adiciona disciplinas que o aluno original ainda não cursou
            for disciplina in historico:
                if disciplina not in aluno_original and disciplina not in recomendacoes:
                    recomendacoes.append(disciplina)

    return recomendacoes


# Históricos dos alunos
historicos = [
    ["ADS101", "MAT101", "BD101"],
    ["ADS101", "PROG201", "WEB101"],
    ["MAT101", "ENG101", "GEST101"],
    ["BD101", "PROG201", "IA101"]
]

# Histórico do aluno que queremos recomendar
aluno_original = ["ADS101", "MAT101"]

resultado = recomendar_disciplinas(historicos, aluno_original)

print("Disciplinas recomendadas:", resultado)
```

A complexidade da abordagem é O(n × m²)
