```python
#exercício5#
class Pessoa:
    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade

    def faz_aniversario(self):
        self.idade += 1

    def __str__(self):
        return f"{self.nome}, {self.idade} anos"


pessoa1 = Pessoa("Maria", 19)
pessoa2 = Pessoa("João", 20)

print("Antes do aniversário:")
print(pessoa1)
print(pessoa2)

pessoa1.faz_aniversario()
pessoa2.faz_aniversario()

print("\nDepois do aniversário:")
print(pessoa1)
print(pessoa2)
```

```python
#exercício6#
class Lampada:
    def __init__(self):
        self.__estado = False

    def ligar(self):
        self.__estado = True

    def desligar(self):
        self.__estado = False

    def is_acesa(self):
        return self.__estado


lampada = Lampada()

print(lampada.is_acesa())

lampada.ligar()
print(lampada.is_acesa())

lampada.desligar()
print(lampada.is_acesa())
```

```python
#exercicio8#
class Animal:
    def fazer_som(self):
        print("Algum som")


class Cachorro(Animal):
    def fazer_som(self):
        print("Au Au")


class Gato(Animal):
    def fazer_som(self):
        print("Miau")


def animais_falam(lista_animais):
    for animal in lista_animais:
        animal.fazer_som()


animais = [Cachorro(), Gato(), Cachorro()]

animais_falam(animais)
```

```python
#projeto1#
class Livro:
    def __init__(self, titulo, autor, quantidade):
        self.titulo = titulo
        self.autor = autor
        self.__quantidade = quantidade

    def emprestar(self):
        if self.__quantidade > 0:
            self.__quantidade -= 1
            return True
        return False

    def devolver(self):
        self.__quantidade += 1

    def quantidade_disponivel(self):
        return self.__quantidade


class Usuario:
    def __init__(self, nome, matricula):
        self.nome = nome
        self.matricula = matricula
        self.livros_emprestados = []

    def emprestar_livro(self, livro):
        if livro.emprestar():
            self.livros_emprestados.append(livro)
            print(self.nome, "emprestou o livro", livro.titulo)
        else:
            print("Livro indisponível.")

    def devolver_livro(self, livro):
        if livro in self.livros_emprestados:
            livro.devolver()
            self.livros_emprestados.remove(livro)
            print(self.nome, "devolveu o livro", livro.titulo)


livro1 = Livro("Harry Potter", "J.K. Rowling", 2)
usuario1 = Usuario("Maria", "001")

usuario1.emprestar_livro(livro1)

print("Quantidade disponível:", livro1.quantidade_disponivel())

usuario1.devolver_livro(livro1)

print("Quantidade disponível:", livro1.quantidade_disponivel())
```

```python
#projeto2#
class Conta:
    def __init__(self, titular, saldo):
        self.titular = titular
        self._saldo = saldo

    def depositar(self, valor):
        if valor > 0:
            self._saldo += valor
            print("Depósito realizado.")

    def sacar(self, valor):
        if valor > 0 and valor <= self._saldo:
            self._saldo -= valor
            print("Saque realizado.")
        else:
            print("Saldo insuficiente.")

    def mostrar_saldo(self):
        return self._saldo


class ContaCorrente(Conta):
    def __init__(self, titular, saldo, taxa_saque=2):
        super().__init__(titular, saldo)
        self.taxa_saque = taxa_saque

    def sacar(self, valor):
        valor_total = valor + self.taxa_saque

        if valor_total <= self._saldo:
            self._saldo -= valor_total
            print("Saque realizado.")
        else:
            print("Saldo insuficiente.")


class ContaPoupanca(Conta):
    juros = 0.05

    def render_juros(self):
        self._saldo += self._saldo * self.juros
        print("Juros aplicados.")


conta_corrente = ContaCorrente("Maria", 1000)

conta_corrente.depositar(500)
conta_corrente.sacar(100)

print("Saldo da conta corrente:", conta_corrente.mostrar_saldo())


conta_poupanca = ContaPoupanca("João", 1000)

conta_poupanca.render_juros()

print("Saldo da conta poupança:", conta_poupanca.mostrar_saldo())
```
