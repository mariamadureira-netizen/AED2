2. Questão teórica 

A busca binária exige que o vetor esteja ordenado porque o algoritmo compara o valor procurado com o elemento do meio e, a partir disso, decide se deve continuar a busca na metade esquerda ou direita do vetor. Se o vetor não estiver ordenado, essa decisão não será confiável e o algoritmo poderá descartar a parte errada do vetor, não encontrando o elemento mesmo que ele esteja presente.


```python
# 3. Questão prática #
def busca_binaria(vetor, alvo, esq=0, dir=None):
    if dir is None:
        dir = len(vetor) - 1

    if esq > dir:
        return -1

    meio = (esq + dir) // 2

    if vetor[meio] == alvo:
        return meio

    if alvo < vetor[meio]:
        return busca_binaria(vetor, alvo, esq, meio - 1)

    return busca_binaria(vetor, alvo, meio + 1, dir)
```

```python
L = [2, 5, 8, 12, 17, 20, 25]

resultado = busca_binaria(L, 17)

print(resultado)
```

```python
# 4. Questão prática #
def busca_sequencial(vetor, alvo):
    comparacoes = 0

    for i in range(len(vetor)):
        comparacoes += 1

        if vetor[i] == alvo:
            return i, comparacoes

    return -1, comparacoes


def busca_binaria(vetor, alvo):
    esq = 0
    dir = len(vetor) - 1
    comparacoes = 0

    while esq <= dir:
        meio = (esq + dir) // 2
        comparacoes += 1

        if vetor[meio] == alvo:
            return meio, comparacoes

        if alvo < vetor[meio]:
            dir = meio - 1
        else:
            esq = meio + 1

    return -1, comparacoes


L = list(range(1000))

alvo = 2000

resultado_seq, comp_seq = busca_sequencial(L, alvo)
resultado_bin, comp_bin = busca_binaria(L, alvo)

print("Busca sequencial:", resultado_seq)
print("Comparações:", comp_seq)

print("Busca binária:", resultado_bin)
print("Comparações:", comp_bin)
```
