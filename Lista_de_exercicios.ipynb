```python
#exercício 4#

class Livro:
    def __init__(self, titulo, autor, paginas):
        self.titulo = titulo
        self.autor = autor
        self.paginas = paginas

    def descricao(self):
        return f"{self.titulo} — {self.autor} ({self.paginas} páginas)"


livro1 = Livro("Dom Casmurro", "Machado de Assis", 256)
livro2 = Livro("O Pequeno Príncipe", "Antoine de Saint-Exupéry", 96)

print(livro1.descricao())
print(livro2.descricao())
```

```python
#exercício 5#
class Conta:
    def __init__(self, titular):
        self.titular = titular
        self.__saldo = 0

    def depositar(self, valor):
        if valor > 0:
            self.__saldo += valor
        else:
            print("O valor deve ser positivo.")

    def consultar_saldo(self):
        return self.__saldo


conta = Conta("Maria")

conta.depositar(100)
conta.depositar(50)

print("Saldo:", conta.consultar_saldo())
```

```python
#exercício 7#

class Veiculo:
    def __init__(self, placa, velocidade_max):
        self.__placa = placa
        self.__velocidade_max = velocidade_max
        self.__velocidade_atual = 0

    def get_placa(self):
        return self.__placa

    def get_velocidade_max(self):
        return self.__velocidade_max

    def set_velocidade_max(self, nova_max):
        if nova_max > 0 and nova_max >= self.__velocidade_atual:
            self.__velocidade_max = nova_max
        else:
            print("Velocidade máxima inválida.")

    def acelerar(self, incremento=10):
        self.__velocidade_atual += incremento

        if self.__velocidade_atual > self.__velocidade_max:
            self.__velocidade_atual = self.__velocidade_max

    def frear(self, decremento=10):
        self.__velocidade_atual -= decremento

        if self.__velocidade_atual < 0:
            self.__velocidade_atual = 0

    def __str__(self):
        return f"Placa: {self.__placa} | Velocidade: {self.__velocidade_atual} km/h | Máxima: {self.__velocidade_max} km/h"


```

```python
carro = Veiculo("ABC-1234", 100)

print(carro)

carro.acelerar()
carro.acelerar(30)
carro.acelerar(80)

print(carro)

carro.frear(20)
carro.frear(50)

print(carro)
```

```python
#exercício 8#
class Funcionario:
    def __init__(self, nome, salario):
        self.nome = nome
        self.salario = salario

    def calcular_bonus(self):
        return self.salario * 0.05

    def __str__(self):
        return self.nome


class Vendedor(Funcionario):
    def __init__(self, nome, salario, vendas):
        super().__init__(nome, salario)
        self.vendas = vendas

    def calcular_bonus(self):
        return (self.salario * 0.05) + (self.vendas * 0.02)


class Gerente(Funcionario):
    def __init__(self, nome, salario, equipe):
        super().__init__(nome, salario)
        self.equipe = equipe

    def calcular_bonus(self):
        return (self.salario * 0.10) + (len(self.equipe) * 100)
```

```python
funcionarios = [
    Funcionario("João", 3000),
    Vendedor("Maria", 4000, 10000),
    Gerente("Carlos", 6000, ["Ana", "Pedro", "Lucas"])
]
```

```python
for funcionario in funcionarios:
    print(
        funcionario.nome,
        "-",
        type(funcionario).__name__,
        "- Bônus: R$",
        funcionario.calcular_bonus()
    )
```

### Exercício 10 - Sistema de Pedidos de E-commerce

Vamos desenvolver um mini sistema orientado a objetos para pedidos de um e-commerce, integrando os conceitos de POO.

```python
# Classe Produto base
class Produto:
    contador = 0

    def __init__(self, nome, preco):
        Produto.contador += 1
        self.__id = Produto.contador
        self.__nome = nome
        self.__preco = preco

    def get_id(self):
        return self.__id

    def get_nome(self):
        return self.__nome

    def get_preco(self):
        return self.__preco

    def set_preco(self, novo_preco):
        if novo_preco > 0:
            self.__preco = novo_preco
        else:
            print("Preço inválido: deve ser maior que zero.")

    def calcular_frete(self):
        return 0 # Frete padrão para produtos (ou seja, 0 para a base)

    def __str__(self):
        return f"ID: {self.__id} | {self.__nome} | R$ {self.__preco:.2f}"
```

Agora, as classes específicas de produtos: `ProdutoFisico` e `ProdutoDigital`.

```python
# Classe ProdutoFisico
class ProdutoFisico(Produto):
    def __init__(self, nome, preco, peso):
        super().__init__(nome, preco)
        self.peso = peso # peso em kg

    def calcular_frete(self):
        return 5 + (2 * self.peso) # R$ 5,00 fixos + R$ 2,00 por kg

    def __str__(self):
        return f"{super().__str__()} | Peso: {self.peso} kg"
```

```python
# Classe ProdutoDigital
class ProdutoDigital(Produto):
    def __init__(self, nome, preco, tamanho_mb):
        super().__init__(nome, preco)
        self.tamanho_mb = tamanho_mb # tamanho em MB

    def calcular_frete(self):
        return 0 # Produto digital não tem frete

    def __str__(self):
        return f"{super().__str__()} | Tamanho: {self.tamanho_mb} MB"
```

Em seguida, a classe `Cliente`.

```python
# Classe Cliente
class Cliente:
    def __init__(self, nome, email):
        self.__nome = nome
        self.__email = email

    def get_nome(self):
        return self.__nome

    def get_email(self):
        return self.__email

    def __str__(self):
        return f"{self.__nome} ({self.__email})"
```

Por fim, a classe `Pedido` que integrará clientes e produtos.

```python
# Classe Pedido
class Pedido:
    contador = 0

    def __init__(self, cliente):
        Pedido.contador += 1
        self.__numero = Pedido.contador
        self.__cliente = cliente
        self.__produtos = []
        self.__status = "Aberto"

    def adicionar_produto(self, produto):
        if self.__status == "Fechado":
            print("Não é possível adicionar produtos. O pedido está fechado.")
        else:
            self.__produtos.append(produto)
            print(f"Produto '{produto.get_nome()}' adicionado ao pedido {self.__numero}.")

    def calcular_total(self):
        total = 0
        for produto in self.__produtos:
            total += produto.get_preco()
        return total

    def calcular_frete_total(self):
        total_frete = 0
        for produto in self.__produtos:
            total_frete += produto.calcular_frete()
        return total_frete

    def fechar_pedido(self):
        self.__status = "Fechado"
        print(f"Pedido {self.__numero} fechado.")

    def __str__(self):
        texto = f"\n---- Resumo do Pedido {self.__numero} ----\n"
        texto += f"Cliente: {self.__cliente}\n"
        texto += f"Status: {self.__status}\n"
        texto += "Produtos:\n"
        if not self.__produtos:
            texto += "  (Nenhum produto no pedido)\n"
        else:
            for produto in self.__produtos:
                texto += f"  - {produto} (Frete: R$ {produto.calcular_frete():.2f})\n"
        texto += f"Total dos Produtos: R$ {self.calcular_total():.2f}\n"
        texto += f"Total do Frete: R$ {self.calcular_frete_total():.2f}\n"
        texto += f"TOTAL GERAL DO PEDIDO: R$ {(self.calcular_total() + self.calcular_frete_total()):.2f}\n"
        texto += "------------------------------------"
        return texto
```

Agora, vamos demonstrar o funcionamento do sistema:

```python
# Demonstração no programa principal

# 1. Criar um cliente
cliente1 = Cliente("Ana Silva", "ana.silva@email.com")
print(f"Cliente Criado: {cliente1}\n")

# 2. Criar produtos (misturando físicos e digitais)
produto_fisico1 = ProdutoFisico("Smartphone", 1500.00, 0.3) # 0.3 kg
produto_fisico2 = ProdutoFisico("Fone de Ouvido", 250.00, 0.1) # 0.1 kg
produto_digital1 = ProdutoDigital("Ebook Python", 50.00, 10) # 10 MB
produto_digital2 = ProdutoDigital("Curso WebDev", 300.00, 500) # 500 MB

print("Produtos Criados:")
print(produto_fisico1)
print(produto_fisico2)
print(produto_digital1)
print(produto_digital2)
print(f"\nTotal de produtos únicos criados (contador da classe Produto): {Produto.contador}\n")

# 3. Montar um pedido
pedido1 = Pedido(cliente1)
pedido1.adicionar_produto(produto_fisico1)
pedido1.adicionar_produto(produto_digital1)
pedido1.adicionar_produto(produto_fisico2)

# 4. Exibir o resumo do pedido
print(pedido1)

# 5. Fechar o pedido
pedido1.fechar_pedido()

# 6. Tentar adicionar mais um produto após o fechamento
print("\nTentando adicionar um produto após o fechamento do pedido:")
pedido1.adicionar_produto(produto_digital2)

# Exibir o pedido novamente para confirmar que o produto não foi adicionado
print(pedido1)


# Exemplo de setter validado para preço
print("\nTestando setter de preço (ProdutoFisico1):")
print(f"Preço atual do {produto_fisico1.get_nome()}: R$ {produto_fisico1.get_preco():.2f}")
produto_fisico1.set_preco(1400.00)
print(f"Novo preço do {produto_fisico1.get_nome()}: R$ {produto_fisico1.get_preco():.2f}")
produto_fisico1.set_preco(-100.00) # Preço inválido
print(f"Preço do {produto_fisico1.get_nome()} após tentativa inválida: R$ {produto_fisico1.get_preco():.2f}")
```
