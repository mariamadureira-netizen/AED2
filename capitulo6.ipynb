```python
#5.#
def decimal_para_hexa(n):
    pilha = []
    simbolos = "0123456789ABCDEF"

    if n == 0:
        return "0"

    while n > 0:
        pilha.append(simbolos[n % 16])
        n //= 16

    resultado = ""

    while pilha:
        resultado += pilha.pop()

    return resultado


print(decimal_para_hexa(26))
```

```python
#6.#
def balanceia_colchetes(expressao):
    pilha = []
    pares = {')': '(', ']': '[', '}': '{'}

    for caractere in expressao:
        if caractere in "([{":
            pilha.append(caractere)

        elif caractere in ")]}":
            if not pilha or pilha.pop() != pares[caractere]:
                return False

    return len(pilha) == 0


print(balanceia_colchetes("({[]})"))
print(balanceia_colchetes("({[}])"))
```

```python
#projeto1#
class Historico:
    def __init__(self):
        self.pilha = []

    def aplicar_filtro(self, estado):
        self.pilha.append(estado)

    def desfazer(self):
        if len(self.pilha) > 1:
            self.pilha.pop()

    def estado_atual(self):
        return self.pilha[-1] if self.pilha else None


historico = Historico()

historico.aplicar_filtro("Todos os dados")
historico.aplicar_filtro("São Paulo")
historico.aplicar_filtro("Clientes ativos")

print(historico.estado_atual())

historico.desfazer()

print(historico.estado_atual())
```

```python
#projeto2#
def calcular(expressao):
    numeros = []
    operadores = []

    prioridade = {'+': 1, '-': 1, '*': 2, '/': 2}

    def operar():
        op = operadores.pop()
        b = numeros.pop()
        a = numeros.pop()

        if op == '+':
            numeros.append(a + b)
        elif op == '-':
            numeros.append(a - b)
        elif op == '*':
            numeros.append(a * b)
        elif op == '/':
            numeros.append(a / b)

    i = 0

    while i < len(expressao):
        c = expressao[i]

        if c == ' ':
            i += 1
            continue

        if c.isdigit():
            numero = 0

            while i < len(expressao) and expressao[i].isdigit():
                numero = numero * 10 + int(expressao[i])
                i += 1

            numeros.append(numero)
            continue

        if c == '(':
            operadores.append(c)

        elif c == ')':
            while operadores[-1] != '(':
                operar()
            operadores.pop()

        elif c in "+-*/":
            while (operadores and operadores[-1] != '(' and
                   prioridade[operadores[-1]] >= prioridade[c]):
                operar()

            operadores.append(c)

        i += 1

    while operadores:
        operar()

    return numeros[-1]


print(calcular("3 + 4 * (2 - 1)"))
```
