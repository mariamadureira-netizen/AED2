```python
# Quantidade de tweets
tweets = 1_000_000

# Tempos por tweet
tempo_preprocessamento = 1 / 100       # 1 segundo a cada 100 tweets
tempo_dicionario = 0.1                 # 0,1 segundo por tweet
tempo_machine_learning = 0.05         # 0,05 segundo por tweet

# Cálculo dos tempos
preprocessamento = tweets * tempo_preprocessamento
dicionario = tweets * tempo_dicionario
machine_learning = tweets * tempo_machine_learning

# Tempo total
tempo_total = preprocessamento + dicionario + machine_learning

# Conversão para minutos e horas
minutos = tempo_total / 60
horas = minutos / 60

# Exibição dos resultados
print("Quantidade de tweets:", tweets)
print("Pré-processamento:", preprocessamento, "segundos")
print("Análise com dicionário:", dicionario, "segundos")
print("Análise com Machine Learning:", machine_learning, "segundos")
print("Tempo total:", tempo_total, "segundos")
print("Tempo total:", minutos, "minutos")
print("Tempo total:", horas, "horas")

# Complexidade
print("\nComplexidade:")
print("Pré-processamento: O(n)")
print("Análise com dicionário: O(n)")
print("Análise com Machine Learning: O(n)")
print("Complexidade total: O(n)")
```
