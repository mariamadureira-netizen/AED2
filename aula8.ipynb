```python
# ============================================================
# EXERCÍCIO 1 — COMPLEXIDADE E IMPLEMENTAÇÃO
# ============================================================

# I. Inserir no início: O(1) → VERDADEIRO
# II. Buscar elemento: O(n) → VERDADEIRO
# III. Python list é lista encadeada → FALSO
#
# Resposta: B. V, V, F
```

```python
# ============================================================
# EXERCÍCIO 2 — INSERINDO SEMPRE NA CABEÇA
# ============================================================

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    def add_head(self, data):
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    def print_list(self):
        atual = self.head
        elementos = []

        while atual is not None:
            elementos.append(atual.data)
            atual = atual.next

        print(elementos)


# Exemplo
L = LinkedList()

L.add_head(1)
L.add_head(2)
L.add_head(3)

print("Exercício 2:")
L.print_list()

# Resultado: [3, 2, 1]
```

```python
# ============================================================
# EXERCÍCIO 3 — REMOÇÃO DE UM NÓ POR VALOR
# ============================================================

class LinkedListRemove:
    def __init__(self):
        self.head = None

    def remove(self, value):
        atual = self.head
        anterior = None

        while atual is not None:

            if atual.data == value:

                if anterior is None:
                    self.head = atual.next
                else:
                    anterior.next = atual.next

                return

            anterior = atual
            atual = atual.next


# Exemplo
L = LinkedListRemove()

L.head = Node(5)
L.head.next = Node(7)
L.head.next.next = Node(9)
L.head.next.next.next = Node(11)

L.remove(9)

print("\nExercício 3:")
atual = L.head

while atual is not None:
    print(atual.data, end=" ")
    atual = atual.next

print()

# Resultado: 5 7 11
```

```python
# ============================================================
# EXERCÍCIO 4 — COMPLEXIDADE DA BUSCA
# ============================================================

# Em uma lista encadeada simples:
#
# A busca precisa percorrer os nós um por um.
#
# Complexidade: O(n)
#
# Resposta: C. O(n)
```

```python
# ============================================================
# EXERCÍCIO 5 — COMPRIMENTO DE UMA LISTA ENCADEADA
# ============================================================

class LinkedListLength:
    def __init__(self):
        self.head = None

    def length(self):
        contador = 0
        atual = self.head

        while atual is not None:
            contador += 1
            atual = atual.next

        return contador


# Exemplo
L = LinkedListLength()

L.head = Node(5)
L.head.next = Node(7)
L.head.next.next = Node(9)

print("\nExercício 5 - Comprimento:")
print(L.length())

# Resultado: 3
```

```python
# ============================================================
# EXERCÍCIO 6 — ENCONTRANDO O ELEMENTO DO MEIO
# ============================================================

class LinkedListMiddle:
    def __init__(self):
        self.head = None

    def find_middle(self):
        slow_ptr = self.head
        fast_ptr = self.head

        while fast_ptr is not None and fast_ptr.next is not None:
            slow_ptr = slow_ptr.next
            fast_ptr = fast_ptr.next.next

        return slow_ptr.data


# Exemplo
L = LinkedListMiddle()

L.head = Node(5)
L.head.next = Node(7)
L.head.next.next = Node(9)
L.head.next.next.next = Node(11)
L.head.next.next.next.next = Node(13)

print("\nExercício 6 - Elemento do meio:")
print(L.find_middle())

# Resultado: 9
```

```python
# ============================================================
# EXERCÍCIO 7 — REVERTENDO A LISTA ENCADEADA
# ============================================================

class LinkedListReverse:
    def __init__(self):
        self.head = None

    def reverse(self):
        prev = None
        current = self.head

        while current is not None:
            next_node = current.next
            current.next = prev
            prev = current
            current = next_node

        self.head = prev

    def display(self):
        atual = self.head

        while atual is not None:
            print(atual.data, end=" ")
            atual = atual.next

        print()


# Exemplo
L = LinkedListReverse()

L.head = Node(5)
L.head.next = Node(7)
L.head.next.next = Node(9)
L.head.next.next.next = Node(11)

print("\nExercício 7 - Lista original:")
L.display()

L.reverse()

print("Exercício 7 - Lista invertida:")
L.display()

# Resultado:
# Original: 5 7 9 11
# Invertida: 11 9 7 5
```

```python
# Exemplo do Projeto 1

playlist = MusicPlaylist()

playlist.add_song(Music("Música A", "Artista A"))
playlist.add_song(Music("Música B", "Artista B"))
playlist.add_song(Music("Música C", "Artista C"))

print("\n================ PLAYLIST ================")
playlist.display_playlist()

print("\nPróxima música:")
print(playlist.play_next())

print("\nPróxima música:")
print(playlist.play_next())

print("\nMúsica anterior:")
print(playlist.play_previous())

print("\nRemovendo Música B:")
playlist.remove_song("Música B")

print("\nPlaylist atualizada:")
playlist.display_playlist()


# ============================================================
# PROJETO 2 — HISTÓRICO DE NAVEGAÇÃO DO BROWSER
# ============================================================

class BrowserNode:
    def __init__(self, url):
        self.url = url
        self.prev = None
        self.next = None


class BrowserHistory:
    def __init__(self):
        self.head = None
        self.tail = None
        self.current = None

    def visit_page(self, url):

        new_node = BrowserNode(url)

        # Primeira página
        if self.current is None:
            self.head = new_node
            self.tail = new_node
            self.current = new_node
            return

        # Descarta o histórico futuro
        self.current.next = None
        self.tail = self.current

        # Conecta a nova página
        new_node.prev = self.current
        self.current.next = new_node

        # Nova página vira a atual
        self.current = new_node
        self.tail = new_node

    def go_back(self):

        if self.current is not None:
            if self.current.prev is not None:
                self.current = self.current.prev

            return self.current.url

        return None

    def go_forward(self):

        if self.current is not None:
            if self.current.next is not None:
                self.current = self.current.next

            return self.current.url

        return None

    def display_history(self):

        current = self.head

        while current is not None:

            if current == self.current:
                print("> " + current.url)
            else:
                print(current.url)

            current = current.next
```

```python
# Exemplo do Projeto 2

browser = BrowserHistory()

browser.visit_page("google.com")
browser.visit_page("youtube.com")
browser.visit_page("instagram.com")

print("\n============ HISTÓRICO ============")
browser.display_history()

print("\nVoltar:")
print(browser.go_back())

print("\nVoltar novamente:")
print(browser.go_back())

print("\nAvançar:")
print(browser.go_forward())

print("\nVisitando nova página:")
browser.visit_page("wikipedia.org")

print("\nHistórico atualizado:")
browser.display_history()

print("\nTentando avançar:")

resultado = browser.go_forward()

if resultado:
    print(resultado)
else:
    print("Não existe página para avançar.")
```
