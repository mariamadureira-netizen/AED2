Projeto 1 

```python
contatos = {
    "Ana": {
        "telefone": "1234-5678",
        "email": "ana@email.com"
    },
    "Bruno": {
        "telefone": "9876-5432",
        "email": "bruno@email.com"
    }
}

# Adicionar um novo contato
contatos["Carlos"] = {
    "telefone": "1122-3344",
    "email": "carlos@email.com"
}

# Buscar um contato pelo nome
nome = "Ana"

if nome in contatos:
    print("Nome:", nome)
    print("Telefone:", contatos[nome]["telefone"])
    print("E-mail:", contatos[nome]["email"])
else:
    print("Contato não encontrado.")

# Listar todos os contatos
print("\nTodos os contatos:")

for nome, dados in contatos.items():
    print(nome, "-", dados["telefone"], "-", dados["email"])
```

```python
tarefas = ["Estudar Python", "Fazer exercícios", "Ler um livro"]

# Adicionar uma nova tarefa
tarefas.append("Organizar o quarto")

# Marcar uma tarefa como concluída
tarefas.remove("Fazer exercícios")

# Visualizar as tarefas pendentes
print("Tarefas pendentes:")

for tarefa in tarefas:
    print("-", tarefa)
    
```
